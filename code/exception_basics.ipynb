```python
# Why Use Exceptions?
    # four main statements for exception handling:
    # - try/except/else/finally – Catch and handle errors.
    # - raise – Manually generate an exception.
    # - assert – Generate an exception if a condition is false.
    # - with – Use context managers to manage resources and handle cleanup automatically.
# Exceptions allow a program to stop its current operation and jump directly to an error handler, instead of continuing with code that cannot succeed.

# Exception Roles
    # Exceptions have five main purposes in Python:
    # 1. Error handling – Catch errors and recover instead of letting the program stop.
    # 2. Event notification – Signal that something happened, without using return codes or flags.
    # 3. Special-case handling – Handle rare or unusual situations in one place instead of checking them everywhere.
    # 4. Termination actions – Use finally or with to ensure resources (files, connections, etc.) are properly closed, whether an error occurs or not.
    # 5. Unusual control flows – Use exceptions to jump out of multiple loops or implement advanced control flows.

# Exceptions: The Short Story
# Default Exception Handler
def fetcher(obj, index):
    return obj[index]

food = 'pizza'
fetcher(food, 4) # Like x[4], last item # 'a'
fetcher(food, 5) # Default handler – console interface # IndexError: string index out of range

    # REPL / Console / IDLE: Displays the error and traceback but allows you to continue entering commands.
    # Normal Python script: Displays the error and traceback, then terminates the program.

# Catching Exceptions
# 1. Basic exception handling
try:
    fetcher(food, 5)
except IndexError:          # Catch and recover
    print('got exception')  # got exception # The error is handled instead of terminating the program.

# 2. Continuing execution after an exception
def catcher():
    try:
        fetcher(food, 5)
    except IndexError:
        print('got exception') # Catch and recover more
    print('continuing')
catcher()  # got exception # continuing

# 3. Can Python return to the line that caused the exception?
    # No. Once an exception is caught:
# 4. Does Python undo changes made before an exception?
    # No. Python does not automatically roll back changes made before the exception.

L, S = [], 'text'
def modder():
    L.append('added')       # Change a mutable
    global S; S = 'changed' # Change a global
    fetcher(food, 5)        # Trigger an exception

try:
    modder()
except IndexError:
    print('got exception')

L, S                        # Changes retained, # got exception # (['added'], 'changed')

# Why?
# - L.append('added') modifies a mutable list. The change remains.
# - S = 'changed' modifies a global variable. The change remains.
# - fetcher(food, 5) raises an exception, but Python does not undo previous changes.

# Raising Exceptions
# we can also raise exceptions manually using the raise statement.
# 1. Raising an exception manually
try:
    raise IndexError # Trigger exception manually
except IndexError:
    print('got exception') # got exception

# 2. Raising an exception without catching it
raise IndexError
# 3. Using assert to raise exceptions
# The assert statement can also trigger an exception, but only when its condition is false.
assert 1 < 0, 'Not in this universe!' # AssertionError: Not in this universe!
# 4. Exception chaining (raise from)
# Python also supports raise from, which allows one exception to be connected to another exception.

# User-Defined Exceptions
# created using classes that inherit from a built-in exception class, usually Exception.
# 1. Creating a custom exception
class Combust(Exception): pass  # User-defined exception

def makePizza():
    raise Combust()             # Raise an instance

try:
    makePizza()
except Combust:                 # Catch class name
    print('got exception')      # got exception
    # How it works:
    # 1. class Combust(Exception) creates a new exception class.
    # 2. raise Combust() creates and raises an instance of that exception.
    # 3. except Combust catches that exception.
    # 4. The handler prints the message.
# 2. Customizing the exception message
# You can customize the message displayed when your exception is not caught by defining the __str__() method
class Combust(Exception):
    def __str__(self):
        return 'Call the fire department!...'

raise Combust()                 # Combust: Call the fire department!...

# Termination Actions
# Python provides the finally block in a try statement. Its main purpose is to execute cleanup or
# termination actions regardless of whether an exception occurs or not.
# 1. Basic try/finally
try:
    fetcher(food, 4)
finally: # Termination actions
    print('after fetch')  # 'a' # after fetch
    # try executes the function.
    # finally always executes after the try block finishes.
# 2. What happens when an exception occurs?
def after():
    try:
        fetcher(food, 5)
    finally:
        print('after fetch')    # Always run
    print('after try?')         # Run only if no exception

after()                         # after fetch # IndexError: string index out of range
    # fetcher(food, 5) raises an IndexError.
    # Python executes the finally block and prints after fetch.
    # The exception continues propagating.
    # print('after try?') never executes because the exception prevents normal continuation
# 3. What happens without an exception?
def after():
    try:
        fetcher(food, 4)
    finally:
        print('after fetch') # Both run if no exception
    print('after try?')

after()     # after fetch # after try?

# 4. except vs. finally
    # - except – Catches exceptions and allows you to handle or recover from them.
    # - finally – Executes cleanup actions regardless of whether an exception occurs or is caught.
    # You can combine both in the same try statement.
# 5. Using with as an alternative
# Python also provides the with statement, which can automatically manage cleanup for supported objects.
with open('pizzarobot.txt', 'w') as file:   # Always close file on exit
    file.write('Catch fire!\n')
    # Here:
    # - open() opens the file.
    # - file.write() writes text into it.
    # - When the with block ends, Python automatically closes the file, even if an exception occurs.

    #   Difference:
    # - try/finally is a general-purpose way to guarantee cleanup.
    # - with provides a simpler alternative for objects that support context management. It can also perform startup actions.
```
